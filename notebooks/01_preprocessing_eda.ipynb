# Task 1: Time-Series Preprocessing and Exploratory Analysis

**Dataset:** [Energy consumption, generation, prices and weather](https://www.kaggle.com/datasets/nicholasjhana/energy-consumption-generation-prices-and-weather) (Kaggle), hourly data for Spain from 2015 to 2018.
**Target:** `total load actual`, the hourly electricity demand in MW.

## Problem and dataset
The goal is to forecast Spain's hourly electricity demand from its own recent history and from the weather. Grid operators have to match supply to demand every hour, so forecast errors are costly.

I chose this dataset because it has a clear hourly timestamp, a meaningful numeric target, and many other variables measured over time (generation by source, prices, and weather for five cities). Demand also has daily, weekly and yearly patterns, which makes it a good fit for lagged features and moving averages.

The data comes as two files: `energy_dataset.csv` (one row per hour) and `weather_features.csv` (one row per city per hour).

## Preprocessing functions
The cleaning steps are written as small functions so the same pipeline can be applied again later. I explain each step in the sections below, where it is first used.

```python
TARGET = "total load actual"
CITIES = ["Barcelona", "Bilbao", "Madrid", "Seville", "Valencia"]
WEATHER_VARS = ["temp", "humidity", "pressure", "wind_speed", "clouds_all", "rain_1h"]
LOCAL_TZ = "Europe/Madrid"

# energy columns that are completely empty or always 0
EMPTY_OR_ZERO_COLS = [
    "generation fossil coal-derived gas", "generation fossil oil shale",
    "generation fossil peat", "generation geothermal",
    "generation hydro pumped storage aggregated", "generation marine",
    "generation wind offshore", "forecast wind offshore eday ahead",
]

# generation columns where a 0 is a reporting dropout, not real output
# (each has only 1-4 zero hours in four years)
ZERO_IS_DROPOUT_COLS = [
    "generation biomass", "generation fossil gas", "generation fossil hard coal",
    "generation fossil oil", "generation hydro run-of-river and poundage",
    "generation hydro water reservoir", "generation nuclear", "generation other",
    "generation other renewable", "generation solar", "generation waste",
    "generation wind onshore",
]


def fill_gaps(frame, fill):
    """Close gaps in a time-indexed frame.

    fill='both' draws a straight line between the hours on either side.
    fill='past' repeats the last observed value, so a gap does not depend on
    the hours after it. A gap at the very start has no earlier value, and those
    hours take the next observed one.
    """
    if fill == "both":
        return frame.interpolate(method="time", limit_direction="both")
    if fill == "past":
        return frame.ffill().bfill()
    raise ValueError("fill must be 'both' or 'past'")


def clean_energy(raw, fill="both"):
    """Parse time to UTC, drop useless columns, treat dropout zeros as missing, fill gaps."""
    df = raw.copy()
    df["time"] = pd.to_datetime(df["time"], utc=True)
    df = df.sort_values("time", kind="stable").drop_duplicates("time").set_index("time")
    df = df.drop(columns=EMPTY_OR_ZERO_COLS)
    df[ZERO_IS_DROPOUT_COLS] = df[ZERO_IS_DROPOUT_COLS].mask(df[ZERO_IS_DROPOUT_COLS] == 0)
    return fill_gaps(df, fill)


def clean_weather(raw, fill="both"):
    """Remove duplicates and impossible values, convert to Celsius, one row per hour."""
    w = raw.copy()
    w["city_name"] = w["city_name"].str.strip()
    w["dt_iso"] = pd.to_datetime(w["dt_iso"], utc=True)
    w = w.sort_values(["city_name", "dt_iso"], kind="stable")
    w = w.drop_duplicates(["city_name", "dt_iso"], keep="first")

    w.loc[~w["pressure"].between(900, 1100), "pressure"] = np.nan   # hPa
    w.loc[w["wind_speed"] > 50, "wind_speed"] = np.nan              # m/s
    w["temp"] = w["temp"] - 273.15                                  # Kelvin to Celsius

    wide = []
    for city, g in w.groupby("city_name"):
        g = fill_gaps(g.set_index("dt_iso")[WEATHER_VARS], fill)
        g.columns = [f"{c}_{city}" for c in WEATHER_VARS]
        wide.append(g)
    wide = pd.concat(wide, axis=1)
    for c in WEATHER_VARS:                                          # average over the five cities
        wide[f"{c}_mean"] = wide[[f"{c}_{city}" for city in CITIES]].mean(axis=1)
    wide.index.name = "time"
    return wide


def merge_energy_weather(energy, weather):
    """Join on the hourly UTC timestamp."""
    return energy.join(weather, how="left")


def add_time_features(df, tz=LOCAL_TZ):
    """Calendar features in local time, since demand follows the local clock."""
    df = df.copy()
    local = df.index.tz_convert(tz)
    df["hour"] = local.hour
    df["dayofweek"] = local.dayofweek          # 0 = Monday
    df["month"] = local.month
    df["year"] = local.year
    df["is_weekend"] = (local.dayofweek >= 5).astype(int)
    return df


def add_lag_features(df, target=TARGET, lags=(1, 24, 168), windows=(24, 168)):
    """Lagged load and moving averages. The averages use shift(1) so a row only
    uses information up to the previous hour."""
    df = df.copy()
    for lag in lags:
        df[f"load_lag_{lag}"] = df[target].shift(lag)
    for win in windows:
        df[f"load_ma_{win}"] = df[target].shift(1).rolling(win).mean()
    return df
```

## A. Understanding the dataset

```python
energy_raw = pd.read_csv(DATA_DIR / "energy_dataset.csv")
weather_raw = pd.read_csv(DATA_DIR / "weather_features.csv")
print("energy_dataset.csv  :", energy_raw.shape)
print("weather_features.csv:", weather_raw.shape)
display(energy_raw.head(3))
display(weather_raw.head(3))
```

### Time range and frequency
The timestamps in the energy file include a UTC offset that changes with daylight saving. I converted them to UTC to get one continuous clock, and only use local time later for hour of day and weekday.

```python
t = pd.to_datetime(energy_raw["time"], utc=True).sort_values()
steps = t.diff().dropna().value_counts()
expected = int((t.max() - t.min()) / pd.Timedelta(hours=1)) + 1

print(f"First timestamp (UTC): {t.min()}   | local: {t.min().tz_convert(LOCAL_TZ)}")
print(f"Last  timestamp (UTC): {t.max()}   | local: {t.max().tz_convert(LOCAL_TZ)}")
print(f"Span: {(t.max() - t.min()).days} days (~{(t.max() - t.min()).days / 365.25:.2f} years)")
print(f"Rows: {len(t):,} | expected for a complete hourly series: {expected:,}")
print("Gaps between consecutive timestamps:")
print(steps.to_string())
print("Duplicate timestamps:", t.duplicated().sum())

wt = pd.to_datetime(weather_raw["dt_iso"], utc=True)
print(f"\nWeather range (UTC): {wt.min()} -> {wt.max()}")
```

The energy data is a complete hourly series covering four years. It has 35,064 rows, every step is exactly one hour, and there are no duplicate timestamps, so no resampling is needed. The weather file covers the same period.

### Structure of the weather file
The weather file is in long format with five cities, so there are about five rows per hour. Before merging it with the energy data I checked the city names and repeated (city, hour) rows.

```python
w = weather_raw.assign(city_name=weather_raw["city_name"].str.strip(),
                       dt_iso=pd.to_datetime(weather_raw["dt_iso"], utc=True))
print("Raw city labels:", sorted(weather_raw["city_name"].unique()), "(note the leading space in ' Barcelona')")
print("\nRows per city (raw):")
print(w["city_name"].value_counts().to_string())

dup_mask = w.duplicated(["city_name", "dt_iso"], keep=False)
print(f"\nRows involved in duplicate (city, hour) pairs: {dup_mask.sum():,}")
print(f"Surplus rows to remove: {w.duplicated(['city_name', 'dt_iso']).sum():,}")

num_cols = ["temp", "pressure", "humidity", "wind_speed", "rain_1h", "clouds_all"]
same = w[dup_mask].groupby(["city_name", "dt_iso"])[num_cols].nunique().max(axis=1)
print(f"Duplicate pairs whose numeric readings are all identical: {(same == 1).mean():.1%}")

example = w[dup_mask].sort_values(["city_name", "dt_iso"]).head(4)
display(example[["city_name", "dt_iso", "temp", "humidity", "weather_main", "weather_description"]])
```

Some hours appear more than once for the same city. The numeric readings are identical and only the text description differs (for example "light rain" and "drizzle"), so I kept the first row. After this every city has exactly 35,064 rows, matching the energy file. I also removed the leading space in " Barcelona".

### Missing values

```python
miss = energy_raw.isna().sum()
miss = miss[miss > 0].to_frame("missing_count")
miss["missing_pct"] = (miss["missing_count"] / len(energy_raw) * 100).round(3)
display(miss.sort_values("missing_count", ascending=False))

all_zero = [c for c in energy_raw.columns[1:] if energy_raw[c].dropna().eq(0).all() and energy_raw[c].notna().any()]
all_nan = [c for c in energy_raw.columns[1:] if energy_raw[c].isna().all()]
print("Entirely empty columns:", all_nan)
print("Columns that are always 0:", all_zero)
assert sorted(all_nan + all_zero) == sorted(EMPTY_OR_ZERO_COLS)
```

Two columns are completely empty and six are always 0, so they carry no information and I dropped them (8 columns). The other generation columns are missing only 17 to 19 hours each, and the target `total load actual` is missing 36 hours (0.10%).

For the plots and the questions below I filled these gaps with time-based linear interpolation. The gaps are short and demand changes smoothly from one hour to the next, so the neighbouring hours give a reasonable estimate. Dropping rows would break the equal hourly spacing that lags and moving averages depend on, and filling with the mean would ignore the daily pattern. Forward fill would copy an old value across a peak or a trough, so the figures do not use it. The next cell checks how long the gaps in the target are.


```python
target_raw = energy_raw.assign(time=pd.to_datetime(energy_raw["time"], utc=True)).set_index("time").sort_index()[TARGET]
is_na = target_raw.isna()
run_id = (is_na != is_na.shift()).cumsum()
gap_lengths = is_na.groupby(run_id).sum()
gap_lengths = gap_lengths[gap_lengths > 0]
print("Gap lengths in the target (hours):", sorted(gap_lengths.tolist(), reverse=True))
print(f"Longest gap: {gap_lengths.max()} hours | number of gaps: {len(gap_lengths)}")

energy_clean = clean_energy(energy_raw)
print("\nMissing values after cleaning:", int(energy_clean.isna().sum().sum()))
print("Shape after cleaning:", energy_clean.shape)

# plot the longest gap before and after filling
longest_id = gap_lengths.idxmax()
gap_idx = is_na[run_id == longest_id].index
win = slice(gap_idx[0] - pd.Timedelta(hours=48), gap_idx[-1] + pd.Timedelta(hours=48))
fig, ax = plt.subplots(figsize=(10, 3.6))
ax.plot(energy_clean.loc[win, TARGET], color=ORANGE, lw=2.2, label="After interpolation")
ax.plot(target_raw.loc[win], color=BLUE, lw=1.4, marker="o", ms=3.5, label="Original (gap = missing)")
ax.set_title(f"Longest gap in the target ({gap_lengths.max()} h), filled by a straight line")
ax.set_ylabel("Total load actual (MW)")
ax.legend()
fig.autofmt_xdate()
save(fig, "1A_gap_interpolation")
plt.show()
```

The longest gap is 8 hours and falls on a morning rise in demand. A straight line is a simple fill and cannot reproduce small bumps inside the gap, but only 36 of 35,064 hours (0.1%) are affected, so the impact on the plots is negligible. I used the same method for the generation columns.

The straight line uses the hours on both sides of the gap. A lag that lands inside the gap then depends on the first observed hour after it. The modeling file saved at the end does not do that. It repeats the last observed value, and it marks the 36 hours where the target was missing so those rows can be left out of training. On an 8-hour morning rise that repeats a night value, which is why the plots keep the straight line.

#### Zeros that are really missing values
The summary statistics later show a minimum of 0 MW for nuclear, gas and wind generation, which cannot be real. I counted the zeros in each column to separate genuine zeros from reporting dropouts.


```python
gen_cols = [c for c in energy_clean.columns if c.startswith("generation") or c.startswith("forecast")]
zero_counts = (energy_raw[gen_cols] == 0).sum().to_frame("zero_hours")
zero_counts["share_of_hours_%"] = (zero_counts["zero_hours"] / len(energy_raw) * 100).round(2)
zero_counts["treated_as_dropout"] = zero_counts.index.isin(ZERO_IS_DROPOUT_COLS)
display(zero_counts[zero_counts["zero_hours"] > 0].sort_values("zero_hours", ascending=False))

# when do the dropouts happen, and what was demand then?
dropout_hours = (energy_raw.set_index(pd.to_datetime(energy_raw["time"], utc=True))[ZERO_IS_DROPOUT_COLS] == 0).any(axis=1)
dropout_hours = dropout_hours[dropout_hours].index
print(f"\nHours with at least one dropout zero: {len(dropout_hours)}")
print("Dates:", sorted({d.strftime('%Y-%m-%d') for d in dropout_hours}))
print("Average load in those hours:", round(target_raw.reindex(dropout_hours).mean()), "MW (overall average:", round(target_raw.mean()), "MW)")
print("Zeros left in those columns after cleaning:", int((energy_clean[ZERO_IS_DROPOUT_COLS] == 0).sum().sum()))
```

Twelve generation columns have only 1 to 4 zero hours in four years, and they all fall in just 5 hours on 4 dates when demand was above average. These are reporting dropouts, so I set them to missing and interpolated them. Three columns (brown coal, pumped storage consumption and the solar day-ahead forecast) have thousands of zeros, which are real (idle plants, no pumping, night time), so I left them unchanged.

#### Weather data
The weather file has no missing values, but checking the maximum of each column shows impossible readings.

```python
print("Raw maximum pressure  (hPa):", weather_raw["pressure"].max(), "| plausible range ~ 950-1050")
print("Raw maximum wind speed (m/s):", weather_raw["wind_speed"].max(), "| record gusts in Spain are ~ 50")
print("Pressure outside 900-1100 hPa:", (~weather_raw["pressure"].between(900, 1100)).sum(), "rows")
print("Wind speed above 50 m/s     :", (weather_raw["wind_speed"] > 50).sum(), "rows")
print("Raw temperature range (K)   :", weather_raw["temp"].min(), "to", weather_raw["temp"].max())

fig, axes = plt.subplots(1, 2, figsize=(11, 3.4))
axes[0].boxplot(weather_raw["pressure"], vert=False, flierprops=dict(marker=".", markersize=3, alpha=.4))
axes[0].set_title("Pressure (hPa), raw")
axes[1].boxplot(weather_raw["wind_speed"], vert=False, flierprops=dict(marker=".", markersize=3, alpha=.4))
axes[1].set_title("Wind speed (m/s), raw")
fig.tight_layout()
save(fig, "1A_weather_outliers")
plt.show()
```

The pressure goes up to about 1,000,000 hPa and the wind speed to 133 m/s, which are sensor errors that would distort every statistic and correlation. I set pressure outside 900-1100 hPa and wind speed above 50 m/s to missing and interpolated them within each city. This affects only 48 pressure values and 3 wind values out of 178,396 rows. I also converted the temperature from Kelvin to Celsius.

### Merging weather and energy
`clean_weather` applies the fixes above and puts the weather into one row per hour, with a column for each city and variable (for example `temp_Madrid`) and a five-city average of each variable (for example `temp_mean`). The average gives a single weather signal for the whole country. I kept six variables per city: temperature, humidity, pressure, wind speed, cloud cover and rain in the last hour.

```python
weather_clean = clean_weather(weather_raw)
df_clean = merge_energy_weather(energy_clean, weather_clean)

print("Weather (wide):", weather_clean.shape, "| Merged:", df_clean.shape)
assert len(df_clean) == len(energy_clean) == 35064
assert df_clean.isna().sum().sum() == 0
print("Missing values in the merged dataset:", int(df_clean.isna().sum().sum()))
print("Index is hourly and complete:", (df_clean.index.to_series().diff().dropna() == pd.Timedelta(hours=1)).all())
display(df_clean[[TARGET, "total load forecast", "price actual", "temp_mean", "temp_Madrid", "humidity_mean"]].head())
```

## B. Analytical questions
Before the questions I added calendar features and lagged features. The calendar features use local Spanish time, because demand follows the local clock. The lags are 1, 24 and 168 hours (one hour, one day and one week earlier). The moving averages cover the previous 24 hours and the previous 7 days. They are calculated on the series shifted by one hour, so a row only uses information up to the previous hour and never includes the value it is trying to describe.

The lags in these questions come from the straight-line fill, the same series as the plots. The training file at the end rebuilds them from the past-only fill.


```python
df = add_lag_features(add_time_features(df_clean))
new_cols = [c for c in df.columns if c not in df_clean.columns]
print("Features added:", new_cols)
print("Rows with NaN lags (first week, no history yet):", int(df["load_lag_168"].isna().sum()))
display(df[[TARGET, "load_lag_1", "load_lag_24", "load_lag_168", "load_ma_24", "load_ma_168"]].iloc[[0, 1, 24, 168, 169]].round(0))
DAYS = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
```

### Question 1. Does demand have a long-term trend, and how seasonal is it?
I plotted the daily mean load with a 30-day and a 365-day centred moving average, and the average load for each calendar month.

```python
daily = df[TARGET].tz_convert(LOCAL_TZ).resample("D").mean()
ma30 = daily.rolling(30, center=True).mean()
ma365 = daily.rolling(365, center=True, min_periods=180).mean()

fig, axes = plt.subplots(2, 1, figsize=(11, 8.5), gridspec_kw={"height_ratios": [1.5, 1]})
ax = axes[0]
ax.plot(daily.index, daily, color=GRAY, lw=0.6, alpha=0.6, label="Daily mean")
ax.plot(ma30.index, ma30, color=BLUE, lw=1.8, label="30-day moving average")
ax.plot(ma365.index, ma365, color=ORANGE, lw=2.2, label="365-day moving average")
ax.set_title("Daily electricity demand with moving averages")
ax.set_ylabel("Load (MW)")
ax.legend(loc="upper right", ncol=3)

monthly = df.groupby("month")[TARGET].mean()
axes[1].plot(monthly.index, monthly.values, color=BLUE, lw=2, marker="o", ms=6)
axes[1].set_xticks(range(1, 13), ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])
axes[1].set_ylim(monthly.min() * 0.98, monthly.max() * 1.02)
axes[1].set_title("Average load by calendar month (all years)")
axes[1].set_ylabel("Load (MW)")
fig.tight_layout()
save(fig, "1B_q1_trend_seasonality")
plt.show()

annual = df.groupby("year")[TARGET].mean()
slope_per_day = np.polyfit(np.arange(len(daily)), daily.values, 1)[0]
print("Average load per year (MW):")
print(annual.round(0).to_string())
print(f"\nLinear trend: {slope_per_day * 365:+.0f} MW per year ({slope_per_day * 365 / daily.mean() * 100:+.2f}% of the mean per year)")
print(f"Change 2015 -> 2018: {(annual[2018] / annual[2015] - 1) * 100:+.2f}%")
print("Average load by month (MW):")
print(pd.Series(monthly.round(0).values, index=["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]).to_string())
print(f"Highest month: {monthly.idxmax()} ({monthly.max():.0f} MW) | Lowest month: {monthly.idxmin()} ({monthly.min():.0f} MW)")
print(f"Seasonal swing (highest - lowest month): {monthly.max() - monthly.min():.0f} MW ({(monthly.max() / monthly.min() - 1) * 100:.1f}%)")
```

**Observations**
- There is a slight upward trend. Yearly averages go from 28,363 MW (2015) to 29,065 MW (2018), which is about +2.5% in three years or roughly +190 MW per year. Four years is too short to say whether this continues. The ends of the 365-day curve use partial windows, so they are less reliable.
- Demand peaks twice a year. The 30-day average is high in winter (January 29,793 MW) and in summer (July 29,928 MW, the highest month), and low in spring and autumn (April 27,597 MW, the lowest). This looks like heating in winter and cooling in summer, which I check in Question 5.
- The gap between the highest and lowest month is 2,331 MW (8.4%), which is small compared with the daily swing of about 10,000 MW seen in Question 2.

### Question 2. How does demand change across the hours of the day and the days of the week?
I plotted the average load for every hour and weekday as a heatmap, and compared the average daily profile of weekdays and weekends (local time).

```python
pivot = df.pivot_table(index="hour", columns="dayofweek", values=TARGET, aggfunc="mean")
profile = df.groupby(["is_weekend", "hour"])[TARGET].mean().unstack(0)

fig, axes = plt.subplots(1, 2, figsize=(12.5, 5), gridspec_kw={"width_ratios": [1, 1.2]})
im = axes[0].imshow(pivot.values, aspect="auto", cmap=BLUES, origin="lower")
axes[0].set_xticks(range(7), DAYS)
axes[0].set_yticks(range(0, 24, 3))
axes[0].set_ylabel("Hour of day (local time)")
axes[0].set_title("Mean load by hour and weekday")
axes[0].grid(False)
fig.colorbar(im, ax=axes[0], label="Mean load (MW)")

axes[1].plot(profile.index, profile[0], color=BLUE, lw=2.2, label="Weekday (Mon-Fri)")
axes[1].plot(profile.index, profile[1], color=ORANGE, lw=2.2, label="Weekend (Sat-Sun)")
axes[1].set_xticks(range(0, 24, 3))
axes[1].set_xlabel("Hour of day (local time)")
axes[1].set_ylabel("Mean load (MW)")
axes[1].set_title("Average daily profile")
axes[1].legend()
fig.tight_layout()
save(fig, "1B_q2_daily_weekly_pattern")
plt.show()

wd, we = profile[0], profile[1]
print(f"Weekday: peak {wd.idxmax()}:00 ({wd.max():.0f} MW), trough {wd.idxmin()}:00 ({wd.min():.0f} MW), swing {wd.max() - wd.min():.0f} MW")
print(f"Weekend: peak {we.idxmax()}:00 ({we.max():.0f} MW), trough {we.idxmin()}:00 ({we.min():.0f} MW)")
dow = df.groupby("dayofweek")[TARGET].mean()
print("\nMean load by weekday (MW):")
print(pd.Series(dow.values, index=DAYS).round(0).to_string())
print(f"\nWeekend vs weekday average: {(df.loc[df.is_weekend == 1, TARGET].mean() / df.loc[df.is_weekend == 0, TARGET].mean() - 1) * 100:+.1f}%")
```

**Observations**
- There is a strong daily cycle. On weekdays demand is lowest at 04:00 (22,928 MW), rises to a peak at 13:00 (33,121 MW), dips in the late afternoon, and rises again to a second peak around 21:00. The difference between the night low and the peak is 10,193 MW, which is about 36% of the mean load.
- Weekends have the same night low (22,163 MW) but a much weaker midday peak, and the evening peak at 21:00 (30,480 MW) is the highest point of the day. Weekends are 8.4% lower than weekdays on average. Sunday is the lowest day (26,362 MW) and Monday is the lowest weekday (29,047 MW).
- Hour of day and day of week explain a large part of the variation. Public holidays are not marked in the data and would probably look like Sundays.

### Question 3. Are there lag effects: how strongly is demand related to its own past?
I calculated the autocorrelation of the load for lags from 1 to 336 hours (two weeks) and plotted the load against the load 24 hours earlier.

```python
s = df[TARGET]
lags = np.arange(1, 337)
acf_vals = np.array([s.autocorr(int(l)) for l in lags])

fig, axes = plt.subplots(1, 2, figsize=(12.5, 4.6), gridspec_kw={"width_ratios": [1.6, 1]})
ax = axes[0]
for d in range(24, 337, 24):
    ax.axvline(d, color=GRAY, lw=0.5, alpha=0.4)
ax.plot(lags, acf_vals, color=BLUE, lw=1.6)
low_i = int(acf_vals.argmin())
for lag, offset in ((24, (8, 8)), (168, (8, 8)), (low_i + 1, (8, -14))):
    ax.scatter([lag], [acf_vals[lag - 1]], color=ORANGE, zorder=5, s=36, edgecolor="white")
    ax.annotate(f"lag {lag} h: r = {acf_vals[lag - 1]:.2f}", (lag, acf_vals[lag - 1]),
                textcoords="offset points", xytext=offset, fontsize=9)
ax.set_xlabel("Lag (hours); grey lines mark every 24 h")
ax.set_ylabel("Autocorrelation")
ax.set_title("Autocorrelation of hourly load")

sample = df.dropna(subset=["load_lag_24"]).sample(4000, random_state=42)
axes[1].scatter(sample["load_lag_24"], sample[TARGET], s=6, alpha=0.35, color=BLUE, edgecolor="none")
r24 = df[[TARGET, "load_lag_24"]].corr().iloc[0, 1]
axes[1].set_xlabel("Load 24 h earlier (MW)")
axes[1].set_ylabel("Load now (MW)")
axes[1].set_title(f"Load vs load 24 h earlier (r = {r24:.2f})")
fig.tight_layout()
save(fig, "1B_q3_autocorrelation")
plt.show()

print("Autocorrelation at selected lags:")
for l in [1, 2, 3, 6, 12, 24, 48, 72, 168, 336]:
    print(f"  lag {l:>3} h: {acf_vals[l - 1]:.3f}")
print(f"\nLowest autocorrelation within the first 24 h: {acf_vals[:24].min():.3f} at lag {acf_vals[:24].argmin() + 1} h")

print("Lowest autocorrelation in the two-week window: {:.3f} at lag {} h".format(acf_vals[low_i], low_i + 1))
```

**Observations**
- Demand has a very strong short-term memory. The correlation with the previous hour is 0.95, and it is still 0.83 and 0.68 at 2 and 3 hours back.
- There is a clear 24-hour rhythm. The correlation drops to about zero at 12 hours, then comes back to 0.70 at 24 hours, with a peak at every multiple of 24 hours. The lowest value in the two weeks is -0.293 at lag 231 hours. The troughs sit about 12 to 15 hours away from a multiple of 24, so one of the two hours is daytime and the other is night.
- The weekly effect is smaller. Lag 168 gives 0.66, only slightly above lag 144 (0.65), and two weeks back (336 hours) is still 0.59. These values also include the slow seasonal change from Question 1.
- The lags worth using are 1, 24 and 168 hours. For a day-ahead forecast only lags of 24 hours or more would be available.


### Question 4. Which past-based feature is the best predictor: a lag or a moving average?
I used each feature directly as a simple prediction of the current load and measured the error (MAE, RMSE, MAPE) and the correlation with the actual load. No model is trained here. The operator's own day-ahead forecast is included as a reference.

A moving average on its own mixes night and day, so I also checked it with the hour held fixed. The hour-of-day profile is the mean load at that hour over the whole series. The shift is how far the moving average sits from its own mean at that hour. The weight on the shift is 1, nothing is fitted, and the profile is computed on the same rows it is scored on, so this describes the association on this series rather than error on held-out days.


```python
cmp = df.dropna(subset=["load_lag_168", "load_ma_168"]).copy()
cmp["mean of lag 24 and lag 168"] = (cmp["load_lag_24"] + cmp["load_lag_168"]) / 2

candidates_ = {
    "Last hour (lag 1)": ("load_lag_1", "lag"),
    "Same hour yesterday (lag 24)": ("load_lag_24", "lag"),
    "Same hour last week (lag 168)": ("load_lag_168", "lag"),
    "Mean of lag 24 and lag 168": ("mean of lag 24 and lag 168", "lag"),
    "24-hour moving average": ("load_ma_24", "ma"),
    "7-day moving average": ("load_ma_168", "ma"),
    "Operator day-ahead forecast": ("total load forecast", "ref"),
}
rows = []
for name, (col, kind) in candidates_.items():
    err = cmp[col] - cmp[TARGET]
    rows.append({"predictor": name, "type": kind, "MAE (MW)": err.abs().mean(),
                 "RMSE (MW)": np.sqrt((err ** 2).mean()),
                 "MAPE (%)": (err.abs() / cmp[TARGET]).mean() * 100,
                 "corr with target": cmp[[col, TARGET]].corr().iloc[0, 1]})
res = pd.DataFrame(rows).sort_values("MAE (MW)").reset_index(drop=True)
display(res.round(3))
res.round(3).to_csv(OUT_DIR / "tables" / "q4_predictor_comparison.csv", index=False)

colors = {"lag": BLUE, "ma": ORANGE, "ref": GRAY}
fig, ax = plt.subplots(figsize=(10, 4.6))
r_ = res.iloc[::-1]
ax.barh(r_["predictor"], r_["MAE (MW)"], color=[colors[k] for k in r_["type"]], height=0.62)
for i, v in enumerate(r_["MAE (MW)"]):
    ax.text(v + 25, i, f"{v:,.0f}", va="center", fontsize=9)
ax.set_xlabel("Mean absolute error (MW), lower is better")
ax.set_title("How well does each past-based feature predict the current load?")
ax.legend(handles=[Patch(color=BLUE, label="Lag-based"), Patch(color=ORANGE, label="Moving average"),
                   Patch(color=GRAY, label="Operator forecast (reference)")], loc="upper right")
ax.grid(axis="y", visible=False)
fig.tight_layout()
save(fig, "1B_q4_lag_vs_moving_average")
plt.show()

# Once the hour is known, does the moving average still track high and low demand?
hour_mean = cmp.groupby("hour")[TARGET].transform("mean")
ma24_mean = cmp.groupby("hour")["load_ma_24"].transform("mean")
ma168_mean = cmp.groupby("hour")["load_ma_168"].transform("mean")
shift24 = cmp["load_ma_24"] - ma24_mean
shift168 = cmp["load_ma_168"] - ma168_mean

def _mae(pred):
    return (pred - cmp[TARGET]).abs().mean()

profile_mae = _mae(hour_mean)
profile_ma24_mae = _mae(hour_mean + shift24)
profile_ma168_mae = _mae(hour_mean + shift168)
lag1_shift_mae = _mae(cmp["load_lag_1"] + shift24)
lag24_shift_mae = _mae(cmp["load_lag_24"] + shift24)

within_rows = []
for hour, g in cmp.groupby("hour"):
    within_rows.append({
        "hour": int(hour),
        "corr_ma24": g[TARGET].corr(g["load_ma_24"]),
        "corr_ma168": g[TARGET].corr(g["load_ma_168"]),
    })
within = pd.DataFrame(within_rows).set_index("hour").sort_index()
given_hour = pd.DataFrame({
    "predictor": [
        "Hour-of-day profile",
        "Profile shifted by 24-hour average",
        "Profile shifted by 7-day average",
        "Lag 1 plus the 24-hour shift",
        "Lag 24 plus the 24-hour shift",
    ],
    "MAE (MW)": [profile_mae, profile_ma24_mae, profile_ma168_mae, lag1_shift_mae, lag24_shift_mae],
})
display(within.round(3))
display(given_hour.round(1))
within.round(3).to_csv(OUT_DIR / "tables" / "q4_within_hour_corr.csv")
given_hour.round(1).to_csv(OUT_DIR / "tables" / "q4_ma_given_hour.csv", index=False)

fig, axes = plt.subplots(1, 2, figsize=(12.5, 4.6))
axes[0].plot(within.index, within["corr_ma24"], color=BLUE, lw=2.2, label="24-hour moving average")
axes[0].plot(within.index, within["corr_ma168"], color=ORANGE, lw=2.2, label="7-day moving average")
axes[0].set_xticks(range(0, 24, 2))
axes[0].set_ylim(0, 1)
axes[0].set_xlabel("Hour of day (local time)")
axes[0].set_ylabel("Correlation with load")
axes[0].set_title("Within each hour, correlation with the moving average")
axes[0].legend()

labels = ["Hour-of-day\nprofile", "Shifted by\n24-hour average", "Shifted by\n7-day average"]
vals = [profile_mae, profile_ma24_mae, profile_ma168_mae]
axes[1].bar(labels, vals, color=[GRAY, BLUE, ORANGE], width=0.72)
for i, v in enumerate(vals):
    axes[1].text(i, v + 30, f"{v:,.0f}", ha="center", va="bottom", fontsize=9)
axes[1].set_ylabel("Mean absolute error (MW)")
axes[1].set_ylim(0, max(vals) * 1.12)
axes[1].set_title("Hour profile, shifted by the recent level")
axes[1].grid(axis="x", visible=False)
fig.tight_layout()
save(fig, "1B_q4_ma_given_hour")
plt.show()

print("Mean within-hour correlation: 24-hour average {:.2f}, 7-day average {:.2f}".format(
    within["corr_ma24"].mean(), within["corr_ma168"].mean()))
print("24-hour average, lowest {:.2f} at {:02d}:00, highest {:.2f} at {:02d}:00".format(
    within["corr_ma24"].min(), int(within["corr_ma24"].idxmin()),
    within["corr_ma24"].max(), int(within["corr_ma24"].idxmax())))
print("MAE, hour profile: {:.0f} MW | plus 24-hour shift: {:.0f} MW | plus 7-day shift: {:.0f} MW".format(
    profile_mae, profile_ma24_mae, profile_ma168_mae))
print("MAE, lag 1 plus the same 24-hour shift: {:.0f} MW | lag 24 plus that shift: {:.0f} MW".format(
    lag1_shift_mae, lag24_shift_mae))
```

**Observations**
- The previous hour (lag 1) is the best past-based predictor, with an MAE of 1,064 MW (3.8%), which matches its 0.95 correlation. It is only available one hour ahead.
- Combining lags works better than using either alone. The mean of lag 24 and lag 168 has an MAE of 2,196 MW, compared with 2,451 MW for yesterday and 2,616 MW for last week, about 10% better than lag 24. Each one is wrong in different situations (for example when the day type changes), so averaging cancels part of the error.
- The moving averages are weak predictors on their own: MAE 3,410 MW (correlation 0.46) for 24 hours and 3,814 MW (0.25) for 7 days. Averaging over a day or week removes the daily cycle, so the average cannot tell a 04:00 low from a 13:00 peak.
- With the hour held fixed, the 24-hour average does track the level. Its correlation with load inside each hour averages 0.70, from 0.41 at 07:00 to 0.91 at 21:00. The hour-of-day profile alone has an MAE of 2,394 MW. Shifting that profile by the 24-hour average cuts it to 1,789 MW. The 7-day average is weaker: within-hour correlation 0.40, and the same shift only reaches 2,256 MW.
- Adding that 24-hour shift on top of a lag makes the error worse, 1,064 MW to 2,412 MW for lag 1 and 2,451 MW to 3,370 MW for lag 24. The lag already carries the recent level, and adding the average counts it a second time. The 24-hour average is worth keeping beside the hour of day, not on top of lag 1 or lag 24.
- The operator's day-ahead forecast has an MAE of 316 MW (1.1%), about 7 times better than the best lag rule that would be available a day ahead.


### Question 5. Do weather variables correlate with demand?
Temperature and demand both follow the time of day, so comparing single hours mixes the weather effect with the daily cycle. To avoid this I used daily averages and only weekdays (weekends have a different level, Question 2). The first plot shows the shape of the load and temperature relationship, and the second compares all weather variables.

```python
weather_labels = {"temp_mean": "Temperature", "humidity_mean": "Humidity", "pressure_mean": "Pressure",
                  "wind_speed_mean": "Wind speed", "clouds_all_mean": "Cloud cover", "rain_1h_mean": "Rain (1 h)"}
daily_w = df[[TARGET] + list(weather_labels)].tz_convert(LOCAL_TZ).resample("D").mean()
wk = daily_w[daily_w.index.dayofweek < 5]                       # weekdays only

# average weekday load in 2 degree bands of daily temperature (bands with at least 10 days)
bins = np.arange(0, 36, 2)
band = pd.cut(wk["temp_mean"], bins)
by_band = wk.groupby(band, observed=True)[TARGET].agg(["mean", "size"])
by_band = by_band[by_band["size"] >= 10]
mid = np.array([iv.mid for iv in by_band.index])
comfort = mid[by_band["mean"].values.argmin()]

# distance from the lowest-demand temperature, and the slope on each side of it
wk = wk.assign(temp_dev=(wk["temp_mean"] - comfort).abs())
hot, cold = wk[wk["temp_mean"] > comfort], wk[wk["temp_mean"] <= comfort]
slope_hot = np.polyfit(hot["temp_mean"], hot[TARGET], 1)[0]
slope_cold = np.polyfit(cold["temp_mean"], cold[TARGET], 1)[0]

corr = wk[[TARGET] + list(weather_labels)].corr()[TARGET].drop(TARGET)
corr.index = [weather_labels[i] for i in corr.index]
corr[f"Distance from {comfort:.0f} °C"] = wk[[TARGET, "temp_dev"]].corr().iloc[0, 1]
corr = corr.sort_values()

fig, axes = plt.subplots(1, 2, figsize=(12.5, 4.8), gridspec_kw={"width_ratios": [1.5, 1]})
axes[0].scatter(wk["temp_mean"], wk[TARGET], s=10, alpha=0.4, color=BLUE, edgecolor="none", label="Weekday (daily mean)")
axes[0].plot(mid, by_band["mean"].values, color=ORANGE, lw=2.4, marker="o", ms=4, label="Mean per 2 °C band")
axes[0].axvline(comfort, color=ORANGE, lw=1, ls="--")
axes[0].set_xlabel("Daily mean temperature across 5 cities (°C)")
axes[0].set_ylabel("Daily mean load (MW)")
axes[0].set_title("Weekday load vs temperature")
axes[0].legend(loc="upper center", bbox_to_anchor=(0.5, -0.14), ncol=2)

axes[1].barh(corr.index, corr.values, color=BLUE, height=0.6)
for i, v in enumerate(corr.values):
    axes[1].text(v + (0.02 if v >= 0 else -0.02), i, f"{v:+.2f}", va="center", ha="left" if v >= 0 else "right", fontsize=9)
axes[1].axvline(0, color="#444", lw=0.8)
axes[1].set_xlim(-0.6, 1.0)
axes[1].set_title("Correlation with daily weekday load")
axes[1].grid(axis="y", visible=False)
fig.tight_layout()
save(fig, "1B_q5_weather_vs_load")
plt.show()

hourly_r = df[[TARGET, "temp_mean"]].corr().iloc[0, 1]
print(f"Weekdays analysed: {len(wk)} days | lowest-demand temperature band: about {comfort:.0f} °C ({by_band['mean'].min():.0f} MW)")
print(f"Linear correlation load vs temperature (daily):  {wk[[TARGET, 'temp_mean']].corr().iloc[0, 1]:+.3f}")
print(f"Correlation load vs distance from {comfort:.0f} °C:        {wk[[TARGET, 'temp_dev']].corr().iloc[0, 1]:+.3f}")
print(f"For comparison, hourly linear correlation (day and night mixed): {hourly_r:+.3f}")
print(f"Slope above {comfort:.0f} °C: {slope_hot:+.0f} MW per °C | slope below: {slope_cold:+.0f} MW per °C")
print("\nAverage weekday load by temperature band (MW) and number of days:")
print(by_band.assign(mean=by_band["mean"].round(0)).to_string())
```

**Observations**
- The relationship is V-shaped, not a straight line. Weekday demand is lowest at around 17 °C (about 28,165 MW) and rises on both sides: roughly 254 MW for every degree colder (heating) and 281 MW for every degree hotter (cooling). Days around 7 °C average 30,981 MW and days around 27 °C average 30,781 MW.
- Because of this the plain correlation is misleading. The linear correlation between daily temperature and load is -0.04, since the cold and hot sides cancel out, while the distance from 17 °C gives +0.34. The hourly correlation of +0.21 is also not reliable because it mixes in the day and night cycle.
- Humidity, pressure, wind speed, cloud cover and rain are all between -0.01 and +0.02, so they add almost nothing on their own.
- Even the temperature effect explains only a modest part of the daily variation (about 12%). The wide scatter at every temperature probably comes from public holidays and holiday periods such as August, which are not marked in the data. Heating and cooling degrees around 17 °C would describe the temperature effect better than the raw temperature.

This only suggests that supply conditions matter more than demand, and I did not test it further.


### Question 7. How accurate is the operator's day-ahead forecast, and when does it struggle?
I calculated the mean absolute error of the operator's forecast for each hour of the day and compared it with the simplest day-ahead rule, using the same hour yesterday (lag 24).

```python
cmp2 = df.dropna(subset=["load_lag_24"])
tso_err = (cmp2["total load forecast"] - cmp2[TARGET]).abs()
naive_err = (cmp2["load_lag_24"] - cmp2[TARGET]).abs()
by_hour = pd.DataFrame({"Operator forecast": tso_err.groupby(cmp2["hour"]).mean(),
                        "Same hour yesterday": naive_err.groupby(cmp2["hour"]).mean()})

fig, ax = plt.subplots(figsize=(10, 4.2))
ax.plot(by_hour.index, by_hour["Same hour yesterday"], color=ORANGE, lw=2.2, label="Same hour yesterday (lag 24)")
ax.plot(by_hour.index, by_hour["Operator forecast"], color=BLUE, lw=2.2, label="Operator day-ahead forecast")
ax.set_xticks(range(0, 24, 2))
ax.set_xlabel("Hour of day (local time)")
ax.set_ylabel("Mean absolute error (MW)")
ax.set_title("Forecast error by hour of day")
ax.legend()
fig.tight_layout()
save(fig, "1B_q7_forecast_error_by_hour")
plt.show()

print(f"Operator forecast: MAE {tso_err.mean():.0f} MW | MAPE {(tso_err / cmp2[TARGET]).mean() * 100:.2f}%")
print(f"Same hour yesterday: MAE {naive_err.mean():.0f} MW | MAPE {(naive_err / cmp2[TARGET]).mean() * 100:.2f}%")
print(f"Operator error is largest at {by_hour['Operator forecast'].idxmax()}:00 ({by_hour['Operator forecast'].max():.0f} MW) "
      f"and smallest at {by_hour['Operator forecast'].idxmin()}:00 ({by_hour['Operator forecast'].min():.0f} MW)")
```

**Observations**
- The operator's forecast is accurate and stays steady through the day, with an MAE of 317 MW (1.10%). The error only ranges from 185 MW at 04:00 to 423 MW at 18:00, so it is a little larger when demand is high and changing.
- Using the same hour yesterday is much worse, with an MAE of 2,456 MW (8.64%), about 7.8 times the operator's error. It is worst during the morning rise, peaking near 08:00 at about 3,800 MW, and best at night (about 1,200 MW around 02:00-04:00).
- This gives a reference for how accurate a day-ahead forecast can be: roughly 2,450 MW for the simplest rule and about 317 MW for the operator.

## Summary
- Demand follows a clear daily and weekly pattern (about 10,000 MW between night and peak, with weekends 8.4% lower), a seasonal pattern with winter and summer peaks, and a small upward trend of about 2.5% from 2015 to 2018.
- Demand is strongly related to its own past: the correlation is 0.95 at lag 1, 0.70 at lag 24 and 0.66 at lag 168. Averaging the lag-24 and lag-168 values beats either one alone. A moving average is weak on its own, but shifting the hour-of-day profile by the 24-hour average cuts that profile's error from 2,394 MW to 1,789 MW. The same shift on top of lag 1 or lag 24 makes those rules worse.
- Temperature has a V-shaped effect with the lowest demand around 17 °C. The other weather variables show almost no relationship with demand.
- Price follows demand within the day (+44% from the lowest to the highest demand hours) but not from month to month.
- The operator's day-ahead forecast has an MAE of about 317 MW (1.1%), compared with about 2,450 MW for the same hour yesterday.


### Saving the processed data
`energy_weather_clean.csv` is the hourly series used above, with gaps closed by a straight line.

`energy_weather_features.csv` is the file to train on. Gaps in the load, the generation columns and the weather are filled with the last observed value, and the lags and moving averages are computed from that load. `target_was_missing` is 1 on the 36 hours with no measured load. The target stored there is the previous hour repeated, so those rows should be left out of training. The lag columns are empty for the first week, before a full week of history exists.


```python
target_missing = (
    energy_raw.assign(time=pd.to_datetime(energy_raw["time"], utc=True))
    .sort_values("time", kind="stable")
    .drop_duplicates("time")
    .set_index("time")[TARGET]
    .isna()
)

energy_model = clean_energy(energy_raw, fill="past")
weather_model = clean_weather(weather_raw, fill="past")
model = add_lag_features(add_time_features(merge_energy_weather(energy_model, weather_model)))
model["target_was_missing"] = target_missing.reindex(model.index).fillna(False).astype(int)

df_clean.to_csv(OUT_DIR / "processed" / "energy_weather_clean.csv")
model.to_csv(OUT_DIR / "processed" / "energy_weather_features.csv")

load_diff = (energy_model[TARGET] - energy_clean[TARGET]).abs()
print("Load hours where the past-only fill differs from the straight line:", int((load_diff > 0.1).sum()))
print("Largest difference: {:.0f} MW".format(load_diff.max()))
print("Hours with no measured load:", int(model["target_was_missing"].sum()))
lag_cols = ["load_lag_1", "load_lag_24", "load_lag_168", "load_ma_24", "load_ma_168"]
warmup_missing = int(model[lag_cols].isna().sum().sum())
other_missing = int(model.drop(columns=lag_cols).isna().sum().sum())
print("Missing values in the lag columns (no history yet):", warmup_missing)
print("Missing values everywhere else:", other_missing)
print("Saved:", sorted(p.name for p in (OUT_DIR / "processed").iterdir()))
print("Modeling file:", model.shape, "| Clean file:", df_clean.shape)
```
